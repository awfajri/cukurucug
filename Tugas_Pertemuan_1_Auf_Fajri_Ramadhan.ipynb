# **Studi Kasus: Penentuan Kelompok UKT Mahasiswa Informatika**

```python
import pandas as pd
import numpy as np

np.random.seed(42)
pendapatan = np.round(np.random.uniform(1.0, 30.0, 150), 1)
tanggungan = np.random.randint(1, 6, 150)

kelompok = []
for p, t in zip(pendapatan, tanggungan):
    rasio = p / t
    if rasio <= 1.5: kelompok.append('Kelompok 1')
    elif rasio <= 3.0: kelompok.append('Kelompok 2')
    elif rasio <= 4.5: kelompok.append('Kelompok 3')
    elif rasio <= 6.0: kelompok.append('Kelompok 4')
    elif rasio <= 8.5: kelompok.append('Kelompok 5')
    elif rasio <= 12.0: kelompok.append('Kelompok 6')
    elif rasio <= 18.0: kelompok.append('Kelompok 7')
    else: kelompok.append('Kelompok 8')

data_ukt = pd.DataFrame({
    'Pendapatan_Orang_Tua': pendapatan,
    'Jumlah_Tanggungan': tanggungan,
    'Kelompok_UKT': kelompok
})
print(data_ukt.head())
```

**Penjelasan:**
Jadi diatas ini saya menginisiasi isi dataframe-nya dengan menggunakan fungsi yang disediakan oleh library numpy untuk men-generate angka acak pendapatan dan tanggungan sebanyak 150 baris data, lalu membaginya ke dalam Kelompok 1 sampai 8 berdasarkan rasio pembagiannya, dan mengubahnya menjadi tabel dataframe menggunakan pandas.



---



```python
from sklearn.model_selection import train_test_split

x_pertanyaan = data_ukt[['Pendapatan_Orang_Tua', 'Jumlah_Tanggungan']]
y_jawaban = data_ukt['Kelompok_UKT']

x_train, x_test, y_train, y_test = train_test_split(x_pertanyaan, y_jawaban, test_size=0.3, random_state=42)
```

**Penjelasan:**
Jadi diatas ini saya menyeleksi kolom dari dataframe tersebut menggunakan metode pemotongan dari pandas untuk membagi data menjadi variabel input (x_pertanyaan) dan variabel target (y_jawaban), lalu membaginya lagi menjadi porsi data pelatihan dan pengujian menggunakan fungsi train_test_split dari library sklearn.



---



```python
from sklearn.tree import DecisionTreeClassifier
import sklearn.metrics

mesin_keputusan = DecisionTreeClassifier(criterion='gini', random_state=42)
mesin_keputusan.fit(x_train, y_train)

prediksi = mesin_keputusan.predict(x_test)
print("Akurasi Model: ", sklearn.metrics.accuracy_score(y_test, prediksi))
```

**Penjelasan:**
Jadi diatas ini saya menginisiasi model klasifikasi pohon keputusan menggunakan fungsi DecisionTreeClassifier yang disediakan oleh library sklearn, lalu melatih model tersebut menggunakan fungsi .fit() dengan data latih agar mesin bisa mengenali pola kelompok 1 hingga 8 secara otomatis.



---



```python
from io import StringIO
from sklearn import tree
from IPython.display import Image
import pydotplus

coretan_pohon = StringIO()
tree.export_graphviz(mesin_keputusan, out_file=coretan_pohon,
                     feature_names=['Pendapatan', 'Tanggungan'],
                     class_names=mesin_keputusan.classes_.tolist(),
                     filled=True, rounded=True)

gambar = pydotplus.graph_from_dot_data(coretan_pohon.getvalue())
display(Image(gambar.create_png()))
```

**Penjelasan:**
Jadi diatas ini saya mengubah model klasifikasi yang sudah dilatih tersebut ke dalam format skema grafik menggunakan fungsi export_graphviz, lalu merendernya menggunakan library pydotplus untuk menghasilkan output visual berupa gambar hierarki pohon keputusan yang bercabang sangat panjang ke bawah mencakup ke-8 kelompok UKT tersebut.
